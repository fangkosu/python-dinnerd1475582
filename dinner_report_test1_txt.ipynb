```python
event_name = "  WEEKEND DINNER  "
meal_amount = " NT$1,200 "
people = "4"

service_rate = 0.10
discount = 100
budget_per_person = 300

# 1. 整理聚餐名稱：去除頭尾空白並轉為小寫
clean_event_name = event_name.strip().lower()

# 2. 整理餐費並轉換型別：移除 NT$、逗號與空白，再轉成整數
clean_meal_amount = int(meal_amount.replace("NT$", "").replace(",", "").strip())

# 3. 將文字人數轉成整數
clean_people = int(people)
```

```python
# 計算各項金額
service_fee = clean_meal_amount * service_rate
total_amount = clean_meal_amount + service_fee - discount
fee_per_person = total_amount / clean_people
is_over_budget = fee_per_person > budget_per_person

# 格式化輸出結果
print(f"聚餐名稱：{clean_event_name}")
print(f"原始餐費：{clean_meal_amount:.2f} 元")
print(f"服務費：{service_fee:.2f} 元")
print(f"折價券：{discount:.2f} 元")
print(f"應付總額：{total_amount:.2f} 元")
print(f"分攤人數：{clean_people} 人")
print(f"每人費用：{fee_per_person:.2f} 元")
print(f"每人預算：{budget_per_person:.2f} 元")
print(f"是否超過預算：{is_over_budget}")
```

```python
# 定義報表內容文字
report_content = f"""========================================
        聚餐費用試算報表
========================================
聚餐名稱      ：{clean_event_name}
分攤人數      ：{clean_people} 人
----------------------------------------
原始餐費      ：NT$ {clean_meal_amount:,.2f} 元
服務費 (10%)  ：NT$ {service_fee:,.2f} 元
折價券折抵    ：- NT$ {discount:,.2f} 元
----------------------------------------
應付總金額    ：NT$ {total_amount:,.2f} 元
每人分攤費用  ：NT$ {fee_per_person:,.2f} 元

個人預算上限  ：NT$ {budget_per_person:,.2f} 元
是否超過預算  ：{'⚠️ 是，已超出預算！' if is_over_budget else '🟢 否，在預算內'}
========================================
"""

# 將報表內容寫入 dinner_report.txt 檔案中
file_name = "dinner_report.txt"
with open(file_name, "w", encoding="utf-8") as file:
    file.write(report_content)

print(f"🎉 報表已成功匯出至：{file_name}")
print("\n--- 匯出的報表預覽 ---")
print(report_content)
```
